### Технічні навички

1. Вони повинні бути першим, що кидається в очі.
2. Вони повинні бути релевантними до вакансії. Тобто, якщо ви подаєтесь на AQA python вакансію, мені буде все одно, що у вас є навичка складати шафу з Ікеї. Наприклад, якщо ви напишете, що знаєте Assembler, С#, Jenkins, та працювали з Active Directory - мені це буде корисно, бо в моїх очах цей кандидат буде досвідчений в IT-галузі.

4. Якщо головні навички НЕ співпадають, але у вас є знання зі схожих галузей (для AQA python ви вказуєте, що у вас є навичка AQA C# + Selenium) - це також ок, якщо ви маєте хоч якийсь досвід на python.
5. Що я очікую побачити в вашому CV, коли його читаю:

    4. Будь який з python тест-фреймворків (pytest, unittest, etc…). Звісно буде краще, якщо це буде той самий, який є в вакансії, але підійде будь який, бо принцип роботи в них однаковий

## Python core

### OOP + SOLID

Як на мене, це найважливіша частина інтерв’ю, бо саме ці принципи дозволяють зрозуміти, наскільки глибоко кандидат розуміє концепції програмування, без прив’язки до мови.

Я дуже полюбляю розповідати зв’язану історію. Наприклад, якщо інтерв’юер питає мене про **Абстракцію** - я я буду відповідати приблизно так:

Давайте я одразу розповім про усі принципи ООП та СОЛІД, бо вони дуже пов’язані один з одним. Уявимо, що у нас є клас Трикутника. Цей клас має три властивості - довжини трьох сторон. Завдяки принципу І**нкапсуляції** ми робимо ці поля приватними, та вказуємо, що неможна, наприклад, задати від’ємне значення для сторони трикутника. Також ми хочемо, щоб наш трикутник мав змогу рахувати свою площу та периметр - для цього в ньому будуть 2 методи на підрахунок площі та периметру. Якщо ж ми хочемо, щоб у нас був ще прямокутник, який буде рахувати площу та периметр - у нього будуть такі самі методи, але з іншою реалізацією. При цьому всьому ми можемо, наприклад, додати в список трикутники та прямокутники і викликати циклом **`for`** ці методи - все буде працювати завдяки **поліморфизму**. Щоб спонукати ці класи імплементувати ці методи - створимо класс фігури, яка буде лише декларувати наявність цих методів у усіх нащадках.  ****Ми не можемо створити просто фігуру, це буде чи трикутник, чи прямокутник, чи щось інше, просто фігури не існує, тому робимо цей клас та методи абстрактним, використовуючи принцип **абстракції**, який спонукає нащадків, які з’явились за допомогою **наслідування** імплементувати усі потрібні методи та не дає можливості створити просто фігуру.

Принцип **SingleResponsibility** каже нам, що наш трикутник повинен відповідати тільки за трикутник. Наприклад, якщо метод підрахунку площі буде виводити цю площу в консоль - це порушення цього принципу, бо, можливо, ми хочемо записати її в файл, чи вивести на сайт.

**OpenClosed** принцип каже нам, що ми не повинні змінювати логіку у нашому трикутнику, ми можемо тільки додавати її (наприклад, додати підрахунок кутів), бо ті, хто будуть користуватись нашим модулем і оновлять версію - будуть мати багато проблем, якщо при цьому зміниться логіка їх програми.

**LiskovSubstitution** принцип каже нам, що треба писати класи так, щоб ми могли використовувати будь якого нащадка замість пращура і це не ломало логіку програми. Тобто, якщо ми замість прямокутника підставимо квадрат, який наслідується з прямокутника - все повинно працювати як і раніше.

Останні 2 принципа створювались для більш об’єктно-орієнтовних мов програмування, наприклад Java чи C#, але і у пайтоні можна знайти їм місце. 

**InterfaceSegregation** принцип каже, що можливо, було б краще створити 2 абстрактних класса **`ISquarable`** та **`IPerimeterable`**, які декларують відповідні методи, бо іноді наші фігури будуть потребувати тільки одного методу, а клас фігури буде спонукати нас використовувати обидва методи.

**DependencyInversion** принцип каже, що при розробці модуля ми повинні відштовхуватись від фігури, а не від трикутника, бо тоді наш модуль буде підтримувати будь яку фігуру, навіть поки не існуючу.

В чому різниця між [Patch та Put](https://stackoverflow.com/questions/31089221/what-is-the-difference-between-put-post-and-patch)?

Питання про типи [HTTP кодів](https://en.wikipedia.org/wiki/List_of_HTTP_status_codes).

Точно треба знати хоч взагалі, яка сотня за що відповідає:

- *1xx informational response* – the request was received, continuing process
- *2xx successful* – the request was successfully received, understood, and accepted
- *3xx redirection* – further action needs to be taken in order to complete the request
- *4xx client error* – the request contains bad syntax or cannot be fulfilled
- *5xx server error* – the server failed to fulfil an apparently valid request

З них частіше всього питають про 2xx - все круто, все працює, 4хх - помилка зі сторони клієнту, та 5хх - помилка зі сторони серверу.

Треба бути готовим до того, що більшість того, що вас питаюсь на співбесіді ніколи вам не знадобиться, на жаль, деякі інтерв’юери хочуть просто забити час, чи у них в голові є думка про те, що це необхідно, але, наприклад, за 6+ років досвіду я ніколи не працював з SQL, але на КОЖНІЙ співбесіді мене питали про JOIN-и. Сессія лайв кодінгу на інтерв’ю - це коли вам дают невеличку технічну задачку, яку ви вирішуєте протягом 5-15 хвилин на інтерв’ю.  Зазвичай це маленька алгоритмічна задача. Приклади цих задач ми зможете знайти на [LeetCode](https://leetcode.com/studyplan/top-interview-150/). Також для тренування дуже рекомендую вирішувати задачки на [CodeWars](https://www.codewars.com/), 7-8 кати, не складніше, вам потрібен механічна навичка.

## Technical Assignment

Це більш об’ємне завдання, на 2-15 годин, яке кандидат робе вдома, перед інтерв’ю. І хоча багато людей думають в форматі “Навіщо я це буду робити, якщо мене ще не беруть на роботу?”, для мене - це найкраща можливість показати себе. Я можу спокійно посидіти, витратити більше часу, ніж просять у листі, але висвітити у цьому завданні усі свої сильні сторони. Наприклад, якщо я дуже круто знаю ООП - я зроблю архітектурно правильне рішення, і потім половину часу на інтерв’ю буду розповідати який я крутий (а в цьому випадку на інтерв’ю ви 100% будете розглядати це завдання).

Також ви будете працювати над ним в тих самих умовах, що й далі на проекті. У вас буде усі ресурси інтернету, ChatGPT, і ваш розум і час. При лайв кодінгу ви в стресі та можете зробити помилки, які б не допустили у спокійному стані, тому це також треба пояснити інтерв’юеру.

# Різниця між Українськими інтерв’ю та Європейськими

У цьому розділі ми будемо розмовляти про статистичну можливість зустріти відношення того чи іншого типу. У будь якому місці можна зустріти будь яке інтерв’ю. Але частіше буде так, як описано в цьому розділі.

## Відмінності

### Кількість співбесід

В Україні є більший шанс пройти 1 технічну співбесіду і потрапити на проект вже завтра. Чим далі на захід Європи ви їдете - тим довше буде процес найма, більше кількість співбесід. Наприклад, ваш покірний слуга отримав оффер через 1.5 місяця після першого контакту з HR-ом. І це швидко!

### Cover Letter

Чим далі на захід Європи - тим більше уваги приділяється супровідному листу, тим більше деталей про компанію та чому ви бажаєте в них працювати треба надати.

### Soft Skills vs Hard Skills

Скоріш за все в Україні вам знадобиться трохи більше технічних скілів, ніж в Європі. Але по вимогам до комунікативних навичок - Європа може дати дуже велику фору. Якщо ви співбесідуєтесь на захід (Європа + США):

- У вас немає проблем в житті
- Ви дуже раді бути на цьому інтервью
- Ви посміхаєтесь
- Ви підтримуєте Small-talk
- Ви задаєте питання по проекту
- Ви ВЖЕ ЗНАЄТЕ що робить ця компанія, і ви в захваті від цього
    - Ви прочитали про цю компанію на її сайті.
    - Ви знаєте досягнення компанії і ви в захваті від них (не забудьте це сказати, коли це буде в тему на співбесіді)
    - Коли у вас спитають “Чи є у вас питання до нас?” - у вас вони є. Розпитайте більше про проект, команду, компанію, культуру компанії. Можливо у вас є питання про ваш кар’єрний зріст - саме час їх задати.

Ці правила будуть працювати і в Україні, і вони значно збільшать ваші шанси, але на заході це ВКРАЙ НЕОБХІДНО.

## Схожість

### Soft Skills vs Hard Skills (Це найважливіший пункт усієї програми курсу)

Як би вам не хотілось, але одними технічними навичками ви можете витягнути себе на вакансію тільки якщо ви ДУЖЕ КРУТИЙ УНІКАЛЬНИЙ СПЕЦІАЛІСТ, але щось мені підказує, що такі не ходять на курси для новачків. Тому, це також треба прийняти як аксіому, **УСІ КОМПАНІЇ НАЙМАЮТЬ ЛЮДИНУ**, а не робота-кодера. І ще одне дуже важливе правило - **УСІ КОМПАНІЇ, В ЯКИХ ВИ ХОЧЕТЕ ПРАЦЮВАТИ НАЙМАЮТЬ ПАРТНЕРА**, а не раба.

Що ж витікає з цих правил? А те, що треба поводитись на інтерв’ю так, нібито ви рівні, а не ви шукаєте роботу. Чому так треба поводитись? **ТОМУ ЩО ВИ РІВНІ.** Тому що вони шукають кльового співробітника так само, як ви шукаєте кльову компанію. І тому що, не боятись сказати щось не те, чи неправильно відповісти, чи вступити в суперечку з інтерв’юером, якщо ви вважаєте, що ви праві (А ва ш покірний слуга один раз отримав відмову саме через те, що боявся вступати в суперечку та відстоювати свою позицію. І це була відмова після 2х місяців співбесід, де технічні були вже пройдені) буде простіше, якщо ви спілкуєтесь як рівні (А ви такими і є).

Зазвичай в цьому місці йду коментар типу “Ну казати про впевненість може будь хто, а як і справді вести себе так, нібито ти впевнений в собі?” Треба бути впевненим в собі. Як це зробити? Тільки практика. **Проходження співбесід - це така ж практична навичка, як і кодінг.** Ви же не будете впевненим програмістом, якщо ви не написали ні одної строчки коду? Так само і з співбесідами. Яким би ви крутим технічним спецом не були - це немає нічого спільного з проходженням співбесід. **Ваші технічні скіли - це необхідний (і то не завжди), але недостатній скіл для успішного проходження інтервью.**

# На що викладачі курсу дивляться при найму людей

## Викладач №1 (Той хто писав цю та інші лекції курсу)

Діалог на інтерв’ю. Найважливіше для мене - як людина мислить та спілкується. Будь які технічні навички можна підтягнути (і зробити це доволі швидко). А от якщо вам некомфортно один з одним - робота буде не в радість (а навіщо так працювати?)

Логічне мислення. Повторюсь, технічні навички можна швидко підтягнути. А от робити правильні висновки - складніше. Мені при найму потрібна не правильна відповідь на запитання, а ход мислення, який я можу зрозуміти. Тому, фраза “Я не знаю як це працює, але думаю, що <отак>, тому що <ланцюжок мислення>” буде навіть крутіше ніж просто зазубрена правильна відповідь. Навіть якщо висновок буде неправильний.

Для того, щоб знати правильні відповіді існує ChatGPT, а на роботі мені потрібен колега, який буде думати, генерувати ідеї, сперечатись зі мною, навіть якщо від буде не правий. Це те, що дозволяє робити круті програмні продукти, та те, що вносить світло в наш робочій день.

## Викладач №2 (Той, хто так само писав інші лекції курсу)

Мені важливо, щоб людина отримувала задоволення саме від професії тестувальника - від написання тестів, знаходження багів, складання та аналізу документації. 

- Розгорнута аргументована думка краще ніж коротка.
- Дискутувати краще ніж промовчати.
- Питання, які я від вас чую, і їх якість говорять інколи більше ніж ваші відповіді.

Будь-яка мета досяжна, якщо розбити її на невеличкі етапи і рухатися маленькими кроками. Тут ключове - мати мету і рухатися до неї, а не здаватися, чи відступати назад. У вас усе вийде, якщо ви будет неуклінно йти обраним шляхом.

## Викладач №3 (Той, хто так само писав інші лекції курсу)

Завжди є баланс між роботою та життям. Ми проводимо дуже багато часу на роботі, тому це частина нашого життя - і, мабуть, саме тому ми хочемо бачити поруч з собою людей, які були б як відмінними фахівцями, так і передусім хорошими людьми. Варіанти найму людей зі специфічною психікою - зазвичай рідкісні. В основному всі хочуть бачити на робочих місцях кандидатів, які окрім володіння необхідними профільними знаннями будуть: зрозумілими у тому, що від них очікується, зрозумілими у тому, які estimate вони будуть надавати, зрозумілими у тому, навіть якщо вони допустили помилку - вони не будуть мовчати, а зможуть пояснити, що помилка була допущена з такої-то причини і кандидат готовий зробити все, щоб цiєї помилки не повторилося. 

У багатьох IT компаніях ці прості принципи: Відкритість, Прозорість, Інновації - возвели навіть до культу з тієї простої причини, що дуже комфортно та легко працювати з людьми, від яких зрозуміло, що очікувати.

Навіть якщо ви початківець у AQA, навіть якщо вам важко вписатися в проект - прозорість і відкритість будуть завжди  показниками вашого прагнення бути професіоналом. Старайтеся бути професіоналами і в технічній сфері, і по Soft Skills.

І пам'ятайте “Одна машина може зробити роботу п'яти звичайних людей. Але жодна машина не зробить роботу однiєї незаурядної людини” - без альтернативного погляду, без інновацій  не буде прогресу - завжди шукайте ... і не зупиняйтеся в цьому пошуку: що ще можна автоматизувати, спростити або поліпшити.
