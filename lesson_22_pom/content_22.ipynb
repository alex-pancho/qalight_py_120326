### Принципи Page Object Pattern:

**Page Object** - це паттерн проектування, де кожна веб-сторінка представлена як окремий клас.

**Переваги:**
- ✓ Відокремлення логіки тестів від деталей UI
- ✓ Зменшення дублювання коду
- ✓ Легкість підтримки (зміни UI в одному місці)
- ✓ Покращена читабельність тестів
- ✓ Повторне використання коду

## 5. Написання Page Objects

### 5.1 Base Page

```python
# pages/base_page.py
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.common.exceptions import TimeoutException

class BasePage:
    """Базовий клас для всіх Page Objects"""
    
    def __init__(self, driver):
        self.driver = driver
        self.wait = WebDriverWait(driver, 10)
    
    def find_element(self, locator, timeout=10):
        """Знайти елемент з очікуванням"""
        return WebDriverWait(self.driver, timeout).until(
            EC.presence_of_element_located(locator)
        )
    
    def find_elements(self, locator, timeout=10):
        """Знайти всі елементи з очікуванням"""
        return WebDriverWait(self.driver, timeout).until(
            EC.presence_of_all_elements_located(locator)
        )
    
    def click(self, locator):
        """Клікнути на елемент"""
        element = self.wait.until(EC.element_to_be_clickable(locator))
        element.click()
    
    def type_text(self, locator, text):
        """Ввести текст в поле"""
        element = self.find_element(locator)
        element.clear()
        element.send_keys(text)
    
    def get_text(self, locator):
        """Отримати текст елемента"""
        return self.find_element(locator).text
    
    def is_element_visible(self, locator, timeout=10):
        """Перевірити чи видимий елемент"""
        try:
            WebDriverWait(self.driver, timeout).until(
                EC.visibility_of_element_located(locator)
            )
            return True
        except TimeoutException:
            return False
    
    def scroll_to_element(self, locator):
        """Прокрутити до елемента"""
        element = self.find_element(locator)
        self.driver.execute_script("arguments[0].scrollIntoView();", element)
    
    def get_current_url(self):
        """Отримати поточний URL"""
        return self.driver.current_url
    
    def open_url(self, url):
        """Відкрити URL"""
        self.driver.get(url)
```

### 5.2 Home Page

```python
# pages/home_page.py
from selenium.webdriver.common.by import By
from pages.base_page import BasePage

class HomePage(BasePage):
    """Page Object для головної сторінки"""
    
    # Locators
    SIGNUP_LOGIN_LINK = (By.CSS_SELECTOR, "a[href='/login']")
    PRODUCTS_LINK = (By.CSS_SELECTOR, "a[href='/products']")
    CART_LINK = (By.CSS_SELECTOR, "a[href='/view_cart']")
    CONTACT_LINK = (By.CSS_SELECTOR, "a[href='/contact_us']")
    LOGGED_IN_AS = (By.CSS_SELECTOR, "li > a > b")
    LOGOUT_LINK = (By.CSS_SELECTOR, "a[href='/logout']")
    DELETE_ACCOUNT_LINK = (By.CSS_SELECTOR, "a[href='/delete_account']")
    SUBSCRIPTION_EMAIL = (By.ID, "susbscribe_email")
    SUBSCRIPTION_BUTTON = (By.ID, "subscribe")
    SUCCESS_MESSAGE = (By.CSS_SELECTOR, ".alert-success")
    
    # URL
    URL = "https://automationexercise.com"
    
    def open(self):
        """Відкрити головну сторінку"""
        self.open_url(self.URL)
        return self
    
    def click_signup_login(self):
        """Натиснути на Signup/Login"""
        self.click(self.SIGNUP_LOGIN_LINK)
        from pages.login_page import LoginPage
        return LoginPage(self.driver)
    
    def click_products(self):
        """Перейти до товарів"""
        self.click(self.PRODUCTS_LINK)
        from pages.products_page import ProductsPage
        return ProductsPage(self.driver)
    
    def click_cart(self):
        """Перейти до кошика"""
        self.click(self.CART_LINK)
        from pages.cart_page import CartPage
        return CartPage(self.driver)
    
    def click_contact(self):
        """Перейти до контактної форми"""
        self.click(self.CONTACT_LINK)
        from pages.contact_page import ContactPage
        return ContactPage(self.driver)
    
    def is_logged_in(self):
        """Перевірити чи користувач залогінений"""
        return self.is_element_visible(self.LOGGED_IN_AS)
    
    def get_logged_username(self):
        """Отримати ім'я залогіненого користувача"""
        return self.get_text(self.LOGGED_IN_AS)
    
    def logout(self):
        """Вийти з акаунту"""
        self.click(self.LOGOUT_LINK)
        return self
    
    def delete_account(self):
        """Видалити акаунт"""
        self.click(self.DELETE_ACCOUNT_LINK)
        return self
    
    def subscribe_email(self, email):
        """Підписатись на розсилку"""
        self.type_text(self.SUBSCRIPTION_EMAIL, email)
        self.click(self.SUBSCRIPTION_BUTTON)
        return self
    
    def is_subscription_successful(self):
        """Перевірити успішність підписки"""
        return self.is_element_visible(self.SUCCESS_MESSAGE)
```

### 5.3 Login Page

```python
# pages/login_page.py
from selenium.webdriver.common.by import By
from pages.base_page import BasePage

class LoginPage(BasePage):
    """Page Object для сторінки входу/реєстрації"""
    
    # Login Locators
    LOGIN_EMAIL = (By.CSS_SELECTOR, "input[data-qa='login-email']")
    LOGIN_PASSWORD = (By.CSS_SELECTOR, "input[data-qa='login-password']")
    LOGIN_BUTTON = (By.CSS_SELECTOR, "button[data-qa='login-button']")
    LOGIN_ERROR = (By.CSS_SELECTOR, "form[action='/login'] p")
    
    # Signup Locators
    SIGNUP_NAME = (By.CSS_SELECTOR, "input[data-qa='signup-name']")
    SIGNUP_EMAIL = (By.CSS_SELECTOR, "input[data-qa='signup-email']")
    SIGNUP_BUTTON = (By.CSS_SELECTOR, "button[data-qa='signup-button']")
    SIGNUP_ERROR = (By.CSS_SELECTOR, "form[action='/signup'] p")
    
    # URL
    URL = "https://automationexercise.com/login"
    
    def open(self):
        """Відкрити сторінку login"""
        self.open_url(self.URL)
        return self
    
    def login(self, email, password):
        """Виконати вхід"""
        self.type_text(self.LOGIN_EMAIL, email)
        self.type_text(self.LOGIN_PASSWORD, password)
        self.click(self.LOGIN_BUTTON)
        from pages.home_page import HomePage
        return HomePage(self.driver)
    
    def signup(self, name, email):
        """Почати процес реєстрації"""
        self.type_text(self.SIGNUP_NAME, name)
        self.type_text(self.SIGNUP_EMAIL, email)
        self.click(self.SIGNUP_BUTTON)
        from pages.signup_page import SignupPage
        return SignupPage(self.driver)
    
    def is_login_error_displayed(self):
        """Перевірити чи відображається помилка входу"""
        return self.is_element_visible(self.LOGIN_ERROR)
    
    def get_login_error_message(self):
        """Отримати текст помилки входу"""
        return self.get_text(self.LOGIN_ERROR)
    
    def is_signup_error_displayed(self):
        """Перевірити чи відображається помилка реєстрації"""
        return self.is_element_visible(self.SIGNUP_ERROR)
```

### 5.4 Signup Page

```python
# pages/signup_page.py
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import Select
from pages.base_page import BasePage

class SignupPage(BasePage):
    """Page Object для детальної сторінки реєстрації"""
    
    # Account Information
    TITLE_MR = (By.ID, "id_gender1")
    TITLE_MRS = (By.ID, "id_gender2")
    PASSWORD = (By.ID, "password")
    DAY = (By.ID, "days")
    MONTH = (By.ID, "months")
    YEAR = (By.ID, "years")
    NEWSLETTER = (By.ID, "newsletter")
    SPECIAL_OFFERS = (By.ID, "optin")
    
    # Address Information
    FIRST_NAME = (By.ID, "first_name")
    LAST_NAME = (By.ID, "last_name")
    COMPANY = (By.ID, "company")
    ADDRESS1 = (By.ID, "address1")
    ADDRESS2 = (By.ID, "address2")
    COUNTRY = (By.ID, "country")
    STATE = (By.ID, "state")
    CITY = (By.ID, "city")
    ZIPCODE = (By.ID, "zipcode")
    MOBILE = (By.ID, "mobile_number")
    
    # Button
    CREATE_ACCOUNT_BUTTON = (By.CSS_SELECTOR, "button[data-qa='create-account']")
    
    # Success message
    ACCOUNT_CREATED = (By.CSS_SELECTOR, "h2[data-qa='account-created']")
    CONTINUE_BUTTON = (By.CSS_SELECTOR, "a[data-qa='continue-button']")
    
    def fill_account_details(self, title, password, day, month, year):
        """Заповнити деталі акаунту"""
        if title.lower() == "mr":
            self.click(self.TITLE_MR)
        else:
            self.click(self.TITLE_MRS)
        
        self.type_text(self.PASSWORD, password)
        
        Select(self.find_element(self.DAY)).select_by_value(str(day))
        Select(self.find_element(self.MONTH)).select_by_value(str(month))
        Select(self.find_element(self.YEAR)).select_by_value(str(year))
        
        return self
    
    def subscribe_newsletter(self):
        """Підписатись на newsletter"""
        self.click(self.NEWSLETTER)
        return self
    
    def subscribe_special_offers(self):
        """Підписатись на спецпропозиції"""
        self.click(self.SPECIAL_OFFERS)
        return self
    
    def fill_address_details(self, first_name, last_name, company, address1, 
                            address2, country, state, city, zipcode, mobile):
        """Заповнити адресу"""
        self.type_text(self.FIRST_NAME, first_name)
        self.type_text(self.LAST_NAME, last_name)
        self.type_text(self.COMPANY, company)
        self.type_text(self.ADDRESS1, address1)
        self.type_text(self.ADDRESS2, address2)
        
        Select(self.find_element(self.COUNTRY)).select_by_visible_text(country)
        
        self.type_text(self.STATE, state)
        self.type_text(self.CITY, city)
        self.type_text(self.ZIPCODE, zipcode)
        self.type_text(self.MOBILE, mobile)
        
        return self
    
    def click_create_account(self):
        """Натиснути Create Account"""
        self.scroll_to_element(self.CREATE_ACCOUNT_BUTTON)
        self.click(self.CREATE_ACCOUNT_BUTTON)
        return self
    
    def is_account_created(self):
        """Перевірити чи створено акаунт"""
        return self.is_element_visible(self.ACCOUNT_CREATED)
    
    def click_continue(self):
        """Продовжити після створення акаунту"""
        self.click(self.CONTINUE_BUTTON)
        from pages.home_page import HomePage
        return HomePage(self.driver)
```

### 5.5 Products Page

```python
# pages/products_page.py
from selenium.webdriver.common.by import By
from pages.base_page import BasePage

class ProductsPage(BasePage):
    """Page Object для сторінки товарів"""
    
    # Locators
    SEARCH_INPUT = (By.ID, "search_product")
    SEARCH_BUTTON = (By.ID, "submit_search")
    PRODUCTS_LIST = (By.CSS_SELECTOR, ".features_items .col-sm-4")
    PRODUCT_NAME = (By.CSS_SELECTOR, ".productinfo p")
    PRODUCT_PRICE = (By.CSS_SELECTOR, ".productinfo h2")
    ADD_TO_CART = (By.CSS_SELECTOR, ".productinfo .add-to-cart")
    VIEW_PRODUCT = (By.CSS_SELECTOR, ".choose a")
    CONTINUE_SHOPPING = (By.CSS_SELECTOR, "button.btn-success")
    VIEW_CART_MODAL = (By.CSS_SELECTOR, ".modal-body a")
    
    # URL
    URL = "https://automationexercise.com/products"
    
    def open(self):
        """Відкрити сторінку товарів"""
        self.open_url(self.URL)
        return self
    
    def search_product(self, search_text):
        """Шукати товар"""
        self.type_text(self.SEARCH_INPUT, search_text)
        self.click(self.SEARCH_BUTTON)
        return self
    
    def get_products_count(self):
        """Отримати кількість товарів"""
        return len(self.find_elements(self.PRODUCTS_LIST))
    
    def get_product_names(self):
        """Отримати назви всіх товарів"""
        products = self.find_elements(self.PRODUCT_NAME)
        return [product.text for product in products]
    
    def add_first_product_to_cart(self):
        """Додати перший товар до кошика"""
        products = self.find_elements(self.PRODUCTS_LIST)
        first_product = products[0]
        add_button = first_product.find_element(*self.ADD_TO_CART)
        
        # Scroll and hover
        self.driver.execute_script("arguments[0].scrollIntoView();", first_product)
        self.driver.execute_script("arguments[0].click();", add_button)
        
        return self
    
    def continue_shopping(self):
        """Продовжити покупки"""
        self.click(self.CONTINUE_SHOPPING)
        return self
    
    def view_cart_from_modal(self):
        """Перейти до кошика з модального вікна"""
        self.click(self.VIEW_CART_MODAL)
        from pages.cart_page import CartPage
        return CartPage(self.driver)
    
    def view_product_details(self, index=0):
        """Переглянути деталі товару"""
        products = self.find_elements(self.VIEW_PRODUCT)
        products[index].click()
        from pages.product_detail_page import ProductDetailPage
        return ProductDetailPage(self.driver)
```

### 5.6 Cart Page

```python
# pages/cart_page.py
from selenium.webdriver.common.by import By
from pages.base_page import BasePage

class CartPage(BasePage):
    """Page Object для сторінки кошика"""
    
    # Locators
    CART_ITEMS = (By.CSS_SELECTOR, "#cart_info_table tbody tr")
    PRODUCT_NAME = (By.CSS_SELECTOR, ".cart_description h4 a")
    PRODUCT_PRICE = (By.CSS_SELECTOR, ".cart_price p")
    PRODUCT_QUANTITY = (By.CSS_SELECTOR, ".cart_quantity button")
    TOTAL_PRICE = (By.CSS_SELECTOR, ".cart_total_price")
    DELETE_BUTTON = (By.CSS_SELECTOR, ".cart_quantity_delete")
    PROCEED_TO_CHECKOUT = (By.CSS_SELECTOR, ".btn-default.check_out")
    EMPTY_CART_MESSAGE = (By.CSS_SELECTOR, "#empty_cart")
    
    # URL
    URL = "https://automationexercise.com/view_cart"
    
    def open(self):
        """Відкрити кошик"""
        self.open_url(self.URL)
        return self
    
    def get_cart_items_count(self):
        """Отримати кількість товарів в кошику"""
        try:
            return len(self.find_elements(self.CART_ITEMS))
        except:
            return 0
    
    def get_product_names(self):
        """Отримати назви товарів"""
        names = self.find_elements(self.PRODUCT_NAME)
        return [name.text for name in names]
    
    def get_total_amount(self):
        """Отримати загальну суму"""
        prices = self.find_elements(self.TOTAL_PRICE)
        total = 0
        for price in prices:
            # Remove 'Rs. ' and convert to int
            amount = int(price.text.replace('Rs. ', ''))
            total += amount
        return total
    
    def delete_product(self, index=0):
        """Видалити товар з кошика"""
        delete_buttons = self.find_elements(self.DELETE_BUTTON)
        delete_buttons[index].click()
        return self
    
    def is_cart_empty(self):
        """Перевірити чи кошик порожній"""
        return self.is_element_visible(self.EMPTY_CART_MESSAGE)
    
    def proceed_to_checkout(self):
        """Перейти до оформлення замовлення"""
        self.click(self.PROCEED_TO_CHECKOUT)
        return self
```

```python
# tests/conftest.py
import pytest
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.chrome.options import Options
from pages.home_page import HomePage
from pages.login_page import LoginPage
from utils.test_data import generate_user_data

@pytest.fixture(scope="function")
def driver():
    """Фікстура для створення WebDriver"""
    # Setup
    chrome_options = Options()
    chrome_options.add_argument("--start-maximized")
    chrome_options.add_argument("--disable-notifications")
    # chrome_options.add_argument("--headless")  # Для headless режиму
    
    service = Service(ChromeDriverManager().install())
    driver = webdriver.Chrome(service=service, options=chrome_options)
    driver.implicitly_wait(10)
    
    yield driver
    
    # Teardown
    driver.quit()


@pytest.fixture(scope="function")
def home_page(driver):
    """Фікстура для головної сторінки"""
    page = HomePage(driver)
    page.open()
    return page


@pytest.fixture(scope="function")
def login_page(driver):
    """Фікстура для сторінки входу"""
    page = LoginPage(driver)
    page.open()
    return page


@pytest.fixture(scope="session")
def test_user():
    """Фікстура з тестовими даними користувача"""
    return generate_user_data()


@pytest.fixture(scope="function")
def logged_in_user(driver, test_user):
    """Фікстура для залогіненого користувача"""
    # Спочатку реєструємо користувача
    home_page = HomePage(driver)
    home_page.open()
    
    login_page = home_page.click_signup_login()
    signup_page = login_page.signup(test_user['name'], test_user['email'])
    
    signup_page.fill_account_details(
        test_user['title'],
        test_user['password'],
        test_user['day'],
        test_user['month'],
        test_user['year']
    )
    
    signup_page.fill_address_details(
        test_user['first_name'],
        test_user['last_name'],
        test_user['company'],
        test_user['address1'],
        test_user['address2'],
        test_user['country'],
        test_user['state'],
        test_user['city'],
        test_user['zipcode'],
        test_user['mobile']
    )
    
    signup_page.click_create_account()
    signup_page.click_continue()
    
    yield test_user
    
    # Cleanup - видаляємо акаунт після тесту
    home_page = HomePage(driver)
    home_page.delete_account()
